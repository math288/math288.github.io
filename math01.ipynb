###ベクトルの回転
ベクトルvの回転後のベクトルv'は、二次元かつ反時計回りの場合、以下のような行列をかけることで求めることができる。

$$R =\begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix}$$

この導出を行う。
x座標とy座標を持つ関数のグラフを考え、ある任意の点(a, b) をとる。
その時の点までの距離をr、角度を$\alpha$とする。
すると、 $$a = r\cos\alpha　---①$$
$$b = r\sin\alpha　---②$$
と書くことができる。
$\theta$だけ回転した後の点の座標を(a', b')とすると、
$$a' = r\cos(\alpha + \theta)$$
$$b' = r\sin(\alpha + \theta)$$
となる。
ここで**加法定理**を用いることで以下が導き出せる。
$$a'
= r\cos(\alpha + \theta)
= r(\cos\alpha \cos\theta - \sin\alpha \sin\theta)
=(r\cos\alpha) \cos\theta - (r\sin\alpha)\sin\theta$$
$$b'
= r(\sin\alpha \cos\theta + \cos\alpha \sin\theta=(r\sin\alpha)\cos\theta + (r\cos\alpha)\sin\theta $$
①と②からこれらの式は
$$a' = a\cos\theta - b\sin\theta$$
$$b' = a\sin\theta + b\cos\theta$$
よって、これらを行列とベクトルの積として、移動後のベクトルは
$$\begin{pmatrix} \cos\theta & -\sin\theta \\\sin\theta & \cos\theta\end{pmatrix} \begin{pmatrix} a \\ b\end{pmatrix}
= \begin{pmatrix}  a'\\ b'\end {pmatrix}$$
よって、導出ができた。

```python
v = np.array([1, 0])
theta = np.radians(45) #回転角をラジアンで表示する

R = np.array([
    [np.cos(theta), -np.sin(theta) ],
    [np.sin(theta), np.cos(theta) ]
])
 #回転後のベクトルを計算
v_rotated = R @ v

print(v)
print(v_rotated)
```

###単位行列の計算
単位行列$E$は、ある行列にかけても元の行列は変化しないという性質を持つ。

```python
E = np.array([[1, 0],[0, 1]])
R = np.array([[3, 4],[5, 6]])
print(E @ R)
print(R @ E)
```

実行結果より、単位行列の計算結果が元の行列Rと完全に一致していることがわかる。
よって
$$単位行列E= \begin {pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}$$
と任意の行列
$$R = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$$
の積は
$$RE = ER = R$$
となる。

###転置行列
行列の行と列を入れ替えた行列のことを転置行列という。
$$A = \begin{pmatrix} 1 & 2 \\ 3 & 4 \\ 5 & 6 \end{pmatrix} \implies A^T = \begin{pmatrix} 1 & 3 & 5 \\ 2 & 4 & 6\end{pmatrix}$$

```python
A = np.array([['a', 'b'],['c', 'd'],['e', 'f']])
print(A.T)
print()
print('np.transposeによる実行')
#np.transpose(A)でも可能
print(np.transpose(A))
```

また、以下も成り立つ
$$(AB)^T = B^TA^T$$
しかし、$$(AB^T) = A^TB^T$$
は一般には成り立たない

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print ((A @ B).T)
print (B.T @ A.T)
if ((A @ B).T == (B.T @ A.T)).all():
  print('これらは等しい')
else :
  print ('これらは等しくない')
print()
print('A^TB^Tの計算')
print (A.T @ B.T)
if ((A @ B).T == (A.T @ B.T)).all():
  print('これらも等しい')
else :
  print('これらは等しくない')
```
